## Variables y tipos de datos

Una variable es un nombre que usamos para guardar información. Piensalo de esta forma: una variable es una caja con una etiqueta. La etiqueta es el nombre y dentro de la caja guardamos un valor.

En Python encontraremos principalmente:

- int: números enteros, como 3;
- float: números con punto decimal, como 0.5;
- complex: números complejos, como 1 + 2j;
- str: texto, como "qubit";
- bool: valores lógicos, True o False.

Python reconoce el tipo automáticamente. La función type nos permite preguntarle cuál es.

```python
numero_de_qubits = 2
probabilidad = 0.5
amplitud = 1 + 1j
nombre_del_estado = "Bell" 
esta_normalizado = True

print(numero_de_qubits, type(numero_de_qubits))
print(probabilidad, type(probabilidad))
print(amplitud, type(amplitud))
print(nombre_del_estado, type(nombre_del_estado))
print(esta_normalizado, type(esta_normalizado))
```

## Operaciones básicas

Python puede funcionar como una calculadora. Podemos sumar, restar, multiplicar, dividir, calcular potencias y obtener el residuo de una división. También podemos comparar dos valores.

Es importante distinguir `=` de `==`: el primero asigna un valor y el segundo pregunta si dos valores son iguales.

```python
a = 7
b = 2

print("Suma:", a + b)
print("División:", a / b)
print("Potencia:", a ** b)
print("Residuo:", a % b)
print("¿a es mayor que b?", a > b)
print("¿a es igual a b?", a == b)
```

$$F = 1.8C + 32$$

Vamos a utilizar esta expresión para convertir 25 grados Celsius a Fahrenheit. Guardaremos la temperatura inicial y el resultado en dos variables diferentes.

## Texto y listas

Un str es una cadena de caracteres. Podemos unir cadenas y colocar variables dentro de un texto usando una *f-string*.

Una lista guarda varios valores en orden. Se escribe entre corchetes y sus posiciones comienzan en 0, no en 1. A esas posiciones las llamamos índices.

### Ejemplo 3: Trabajar con una lista

## Condiciones: tomar decisiones

Una condición permite que el programa elija qué hacer. Python utiliza `if`, `elif` y `else`.

Observa la sangría: las líneas que pertenecen a una condición llevan espacios al inicio. En Python esos espacios son parte de la sintaxis.

```python
medicion = 1

if medicion == 0:
    print("Medimos el estado 0")
else:
    print("Medimos el estado 1")

probabilidad = 0.5

if probabilidad > 0.5:
    print("El resultado 1 es más probable")
elif probabilidad < 0.5:
    print("El resultado 0 es más probable")
else:
    print("Ambos resultados son igual de probables")
```

### Ejemplo 4: Comprobar una distribución de probabilidades

Una distribución de probabilidades debe sumar 1. El siguiente código indica que la distribución es válida cuando se cumple esta condición y nos pide revisar las probabilidades en cualquier otro caso.

Usaremos `round(suma, 10)` para evitar pequeños errores de redondeo de la computadora.

```python
probabilidades = [0.25, 0.25, 0.5]
suma = sum(probabilidades)

if round(suma, 10) == 1:
    print("Distribución válida")
else:
    print("Revisar probabilidades")
```

## Ciclos: repetir instrucciones

Un ciclo `for` recorre los elementos de una colección. Esto nos evita copiar la misma instrucción muchas veces.

La función range produce una secuencia de números. Por otro lado, enumerate nos entrega al mismo tiempo la posición y el valor de cada elemento.

```python
mediciones = [0, 1, 1, 0]

for resultado in mediciones:
    print("Medimos:", resultado)

print("---")

for indice, resultado in enumerate(mediciones):
    print(f"Experimento {indice}: resultado {resultado}")

print("---")

for i in range(3):
    print("Repetición", i)
```

### Ejemplo 5: Contar resultados

Vamos a recorrer una lista de mediciones y contar cuántas veces aparece el resultado 1.

Comenzamos el conteo en cero. Cada vez que encontramos un 1, aumentamos el conteo dentro de una condición.

```python
mediciones = [0, 1, 1, 0, 1, 0, 0, 0, 1, 1, 0, 1, 0]
conteo_unos = 0

for resultado in mediciones:
    if resultado == 1:
        conteo_unos = conteo_unos + 1

print("Cantidad de unos:", conteo_unos)
```

## Funciones: crear nuestras propias herramientas

Una función agrupa instrucciones que queremos reutilizar. Se define con `def`, puede recibir valores de entrada y puede devolver un resultado con `return`.

Piensalo de esta forma: una función es una receta. Le damos ingredientes, sigue una serie de pasos y nos entrega un resultado.

```python
def elevar_al_cuadrado(numero):
    resultado = numero ** 2
    return resultado


print(elevar_al_cuadrado(3))
print(elevar_al_cuadrado(0.5))


def describir_medicion(resultado):
    if resultado == 0:
        return "Se midió cero"
    else:
        return "Se midió uno"


print(describir_medicion(1))
```

### Ejemplo 6: Frecuencias de medición

La función calcular_frecuencias recibirá dos conteos y devolverá una lista con sus frecuencias.

Por ejemplo, si recibe 75 ceros y 25 unos debe devolver las frecuencias 0.75 y 0.25.

```python
def calcular_frecuencias(ceros, unos):
    total = ceros + unos
    frecuencia_ceros = ceros / total
    frecuencia_unos = unos / total
    return [frecuencia_ceros, frecuencia_unos]

frecuencias = calcular_frecuencias(75, 25)
print(frecuencias)
```

## Introducción a NumPy

NumPy es una biblioteca de Python para trabajar con arreglos y operaciones numéricas. Durante el laboratorio la usaremos para vectores, matrices, números complejos y constantes como π.

Para trabajar con NumPy la importaremos con el nombre corto np, que es la convención que encontraremos en la mayoría de los programas.

Un arreglo de NumPy se parece a una lista, pero está diseñado para hacer cálculos numéricos de forma directa.

```python
import numpy as np

print("Versión de NumPy:", np.__version__)

lista = [1, 2, 3]
arreglo = np.array([1, 2, 3])

print("Lista por 2:", lista * 2)
print("Arreglo por 2:", arreglo * 2)
print("Tipo del arreglo:", type(arreglo))
```

Podemos operar elemento por elemento, sumar arreglos y calcular su norma. La norma mide el tamaño de un vector. Para un vector `v`, `np.linalg.norm(v)` calcula ese tamaño.

Dividir un vector entre su norma produce un vector normalizado. Esta idea aparecerá constantemente en computación cuántica.

```python
v = np.array([3.0, 4.0])
w = np.array([1.0, 2.0])

print("v + w =", v + w)
print("2v =", 2 * v)
print("Norma de v =", np.linalg.norm(v))

v_normalizado = v / np.linalg.norm(v)
print("v normalizado =", v_normalizado)
print("Norma final =", np.linalg.norm(v_normalizado))
```

### Ejemplo 7: Normalizar un arreglo

En el siguiente ejemplo crearemos un arreglo con dos entradas iguales a 1, calcularemos su norma y lo normalizaremos. Al final comprobaremos que la norma del resultado sea aproximadamente 1.

```python
u = np.array([1.0, 1.0])
norma_u = np.linalg.norm(u)
u_normalizado = u / norma_u

print("Norma original:", norma_u)
print("Arreglo normalizado:", u_normalizado)
print("Norma final:", np.linalg.norm(u_normalizado))
```

## Primera conexión con computación cuántica

Un estado de un qubit puede escribirse como

$$|ψ⟩ = α|0⟩ + β|1⟩,$$

donde α y β pueden ser números complejos y deben satisfacer

$$|α|^2 + |β|^2 = 1.$$

En Python la letra j representa la unidad imaginaria. Con NumPy podemos guardar las amplitudes en un arreglo y obtener las probabilidades calculando el valor absoluto al cuadrado de cada una.

```python
alpha = 1 / np.sqrt(2)
beta = 1j / np.sqrt(2)

estado = np.array([alpha, beta], dtype=complex)
probabilidades = np.abs(estado) ** 2

print("Estado:", estado)
print("Probabilidades [P(0), P(1)]:", probabilidades)
print("Suma de probabilidades:", np.sum(probabilidades))
```

### Ejemplo 8: Normalizar un estado con amplitudes complejas

El arreglo formado por 1 y 1j todavía no representa un estado cuántico normalizado. Para convertirlo en uno, realizaremos los siguientes pasos:

1. calcula su norma;
2. divide el arreglo entre su norma;
3. calcula las probabilidades;
4. verificaremos que sumen 1.

Como las dos amplitudes tienen la misma magnitud, las probabilidades de medir 0 y 1 serán iguales.

```python
estado_sin_normalizar = np.array([1, 1j], dtype=complex)
norma = np.linalg.norm(estado_sin_normalizar)
estado_normalizado = estado_sin_normalizar / norma
probabilidades = np.abs(estado_normalizado) ** 2

print("Norma original:", norma)
print("Estado normalizado:", estado_normalizado)
print("Probabilidades:", probabilidades)
print("Suma:", np.sum(probabilidades))

# Los resultados 0 y 1 tienen la misma probabilidad: 0.5.
```
